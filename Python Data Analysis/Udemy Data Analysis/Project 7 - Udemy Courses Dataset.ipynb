### 10. What are the Max. Number of Subscribers for Each Level of courses ?

```python
df.groupby('level')['num_subscribers'].max()
```

```python
import seaborn as sns
```

```python
graphs= df.groupby('level')['num_subscribers'].max().sort_values(ascending=False)
plt.title('Max Subscribers For Each Course')
plt.xlabel('Level',fontsize = 12,color = 'red')
plt.ylabel('Subscribers',fontsize = 12,color = 'blue')
ax = sns.barplot(x = graphs.index, y = graphs.values)
for bars in ax.containers:
    ax.bar_label(bars)
```
