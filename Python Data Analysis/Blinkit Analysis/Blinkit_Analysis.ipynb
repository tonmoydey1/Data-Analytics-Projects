## *Data Analysis*

Total Sales: The overall revenue generated from all items sold.

```python
sales = df.sales.sum().round(2)
print(f'Total sale is: Rs.{sales}')
```

Average Sales: The average revenue per sale.


```python
avg_sales = df.sales.mean().round(2)
print(f'Average sale is: Rs.{avg_sales}')
```

Number of Items: The total count of different items sold.


```python
no_of_items = df.item_identifier.nunique()
print(f'Number of items are: {no_of_items}')
```

Average Rating: The average customer rating for items sold. 


```python
avg_rating = df.rating.mean().round(2)
print(f'Average rating is: {avg_rating}')
```

Total Sales by Fat Content:


```python
df.head(2)
```

```python
t_s_b_f_c = df.groupby('item_fat_content').sales.sum().round(2)
plt.title('Total Sales By Fat Content',fontweight = 'bold', color = 'Green')
plt.pie(t_s_b_f_c,labels=t_s_b_f_c.index,autopct='%1.1f%%',startangle=90,wedgeprops=dict(width=0.4, edgecolor='white'),colors=['skyblue','lightgreen'])
plt.show()
```

Total Sales by Item Type

```python
df.head(2)
```

```python
df.groupby('item_type')['sales'].sum().round(2).sort_values(ascending=False)
```

```python
t_s_b_i_t = df.groupby('item_type')['sales'].sum().round(2).sort_values(ascending=False)
plt.figure(figsize=(15,6))
plt.title('Total Sales by Item Type')
plt.xlabel('Item Type',fontweight = 'bold')
plt.ylabel('Sales',fontweight = 'bold')
plt.xticks(rotation=90)
ax = sns.barplot(x = t_s_b_i_t.index,y = t_s_b_i_t.values, palette=['cyan','magenta','yellow'])
for i in ax.containers:
    ax.bar_label(i)
plt.plot()
```

Fat Content by Outlet for Total Sales

```python
df.head(2)
```

```python
df.groupby(['item_fat_content','outlet_type'])['sales'].sum().reset_index()
```

```python
fcific = df.groupby(['item_fat_content','outlet_type'])['sales'].sum().reset_index()
plt.figure(figsize=(10,6))
plt.title('Fat Content by Outlet for Total Sales')
ax = sns.barplot(data=fcific,x = 'outlet_type',y='sales',hue = 'item_fat_content',palette=['red','green'])
for i in ax.containers:
    ax.bar_label(i)
plt.tight_layout()
plt.show()
```
