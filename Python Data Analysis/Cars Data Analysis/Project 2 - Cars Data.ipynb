```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
df = pd.read_csv('2. Cars Data1.csv')
```

```python
df
```

```python
df.columns = df.columns.str.lower()
```

1) Instruction ( For Data Cleaning ) 
- Find all Null Value in the dataset. If there is any null value in any column, then fill it with the mean of that column. 

```python
df.isnull().sum()
```

```python
df.make = df.make.fillna(df.make.mode()[0])
```

```python
df.model = df.model.fillna(df.model.mode()[0])
df.type = df.type.fillna(df.type.mode()[0])
df.origin = df.origin.fillna(df.origin.mode()[0])
df.drivetrain = df.drivetrain.fillna(df.drivetrain.mode()[0])
df.msrp =df.msrp.fillna(df.msrp.mode()[0])
df.invoice =df.invoice.fillna(df.invoice.mode()[0])
df.enginesize =df.enginesize.fillna(df.enginesize.mode()[0])
df.cylinders =df.cylinders.fillna(df.cylinders.mode()[0])
df.horsepower =df.horsepower.fillna(df.horsepower.mode()[0])
df.mpg_city =df.mpg_city.fillna(df.mpg_city.mode()[0])
df.mpg_highway =df.mpg_highway.fillna(df.mpg_highway.mode()[0])
df.weight =df.weight.fillna(df.weight.mode()[0])
df.wheelbase =df.wheelbase.fillna(df.wheelbase.mode()[0])
df.length =df.length.fillna(df.length.mode()[0])
```

```python
df.isnull().sum()
```

----

3) Instruction ( Filtering )

4) Instruction ( Removing unwanted records )
- Remove all the records (rows) where Weight is above 4000.

```python
df.drop(df[df.weight > 4000].index,inplace=True)
```

```python
df
```

-----

5) Instruction ( Applying function on a column )
- Increase all the values of 'MPG_City' column by 3.

```python
df.mpg_city = df.mpg_city.apply(lambda x:x+3)
```
