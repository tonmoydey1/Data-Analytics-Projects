### Q.2. In which year highest number of the TV Shows & Movies were released ? Show with Bar Graph.

```python
df.release_date.dtype
```

```python
df['date_n'] = pd.to_datetime(df['release_date'],errors='coerce')
```

```python
graph = df.date_n.dt.year.value_counts()
plt.figure(figsize=(15,5))
plt.title('Highest No. Of TV Shows & Movies')
plt.xlabel('Year')
ax = sns.barplot(x = graph.index,y = graph.values)
for bar in ax.containers:
    ax.bar_label(bar)
```

----

### Q.3. How many Movies & TV Shows are in the dataset ? Show with Bar Graph.

```python
df.head()
```

```python
bar = df.groupby('category').category.count()
plt.title('No, Of Movies & TV shows')
plt.xlabel('Category')
plt.ylabel('Count')
ax = sns.barplot(x = bar.index, y = bar.values)
for i in ax.containers:
    ax.bar_label(i)
```
