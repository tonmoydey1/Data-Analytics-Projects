# The Flights Dataset
***
Here, 
The Flights Booking Dataset of various Airlines is a scraped datewise from a famous website in a structured format. The dataset contains the records of flight travel details between the cities in India. Here, multiple features are present like Source & Destination City, Arrival & Departure Time, Duration & Price of the flight etc.


This data is available as a CSV file. We are going to analyze this data set using the Pandas DataFrame.

This analyse will be helpful for those working in Airlines, Travel domain.

-----------

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('airlines_flights_data.csv')
```

```python
df
```

```python
df.info()
```

# Cleaning the data

```python
df = df.rename(columns={'index':'sl_no'})
```

```python
df
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df.airline.unique()
```

```python
df.airline = df.airline.str.replace('_','' ).str.title()
df.airline = df.airline.str.replace('Spicejet','SpiceJet' )
df.airline = df.airline.str.replace('Airasia','AirAsia' )
df.airline = df.airline.str.replace('Airindia','AirIndia' )
```

```python
df.airline.unique()
```

```python
df.departure_time.unique()
```

```python
df.departure_time = df.departure_time.str.replace( 'Early_Morning',' Early Morning' )
df.departure_time = df.departure_time.str.replace( 'Late_Night','Late Night ')
```

```python
df.departure_time.unique()
```

```python
df.stops.unique()
```

```python
df.stops = df.stops.replace('two_or_more','two or more')
```

```python
df.stops.unique()
```

```python
df.info()
```

```python
df.arrival_time.unique()
```

```python
df.arrival_time = df.arrival_time.str.replace( 'Early_Morning',' Early Morning' )
df.arrival_time = df.arrival_time.str.replace( 'Late_Night','Late Night ')
```

```python
df['class'].unique()
```

```python
df.duration.unique()
```

```python
df.days_left.unique()
```

```python
df.price.unique()
```

## Q.2. `Show Bar Graphs representing the Departure Time & Arrival Time`

```python
DT = df.departure_time.value_counts()
AT = df.arrival_time.value_counts()

plt.title('Representing Departure Time')
plt.xlabel('Departure Time')
plt.ylabel('Count')
ax = sns.barplot(x = DT.index, y= DT.values)
for i in ax.containers:
    ax.bar_label(i)
plt.show()


plt.title('Representing Arrival Time')
plt.xlabel('Arrival Time')
plt.ylabel('Count')
ax = sns.barplot(x = AT.index, y= AT.values)
for j in ax.containers:
    ax.bar_label(j)
plt.show()
```

## Q.3. `Show Bar Graphs representing the Source City & Destination City`

```python
sc = df.source_city.value_counts()
dc = df.destination_city.value_counts()

plt.title('Representing Source City')
plt.xlabel('Source City')
plt.ylabel('Count')
ax = sns.barplot(x = sc.index, y= sc.values)
for i in ax.containers:
    ax.bar_label(i)
plt.show()


plt.title('Representing Destination City')
plt.xlabel(' Destination Time')
plt.ylabel('Count')
ax = sns.barplot(x = dc.index, y= dc.values)
for j in ax.containers:
    ax.bar_label(j)
plt.show()
```

## Q.9. `What will be the Average Price of Vistara airline for a flight from Delhi to Hyderabad in Business Class ?`

```python
df.head(5)
```

```python
avg_price = df[(df.airline == 'Vistara') & (df.source_city == 'Delhi') & (df.destination_city == 'Hyderabad') & (df['class'] == 'Business')].price.mean().round(2)
print(f'Average price of vistara airline is Rs.{avg_price}')
```
