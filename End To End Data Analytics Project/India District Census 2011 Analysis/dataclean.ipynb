## **1. Create a bar chart showing the top 10 states by total population.**

```python
from matplotlib.ticker import FuncFormatter

plt.figure(figsize=(14, 7))
bar_chart = (df.groupby('state_name')['population'].sum().sort_values(ascending=True).tail(10))
colors = sns.color_palette('viridis', n_colors=len(bar_chart))
ax = sns.barplot(x=bar_chart.values, y=bar_chart.index, hue=bar_chart.index, palette=colors, legend=False)

for container in ax.containers:
    ax.bar_label(container, fmt='{:,.0f}', padding=6, fontsize=10, color='#263238')

ax.xaxis.set_major_formatter(FuncFormatter(lambda value, position: f'{value / 1_000_000:.1f}M'))
ax.set_title('Top 10 States by Total Population', fontsize=20, fontweight='bold', pad=18, color='#263238')
ax.set_xlabel('Total Population (millions)', fontsize=12, labelpad=10, color='#455A64')
ax.set_ylabel('')
ax.tick_params(axis='both', labelsize=11, colors='#455A64')
ax.grid(axis='x', linestyle='--', linewidth=0.8, alpha=0.35)
ax.grid(axis='y', visible=False)
ax.set_axisbelow(True)

sns.despine(left=True, bottom=True)
plt.tight_layout()
plt.show()
```

## **2. Create a horizontal bar chart showing the top 10 districts by population, with state labels.**

```python
from matplotlib.ticker import FuncFormatter

top_districts = (df[['district_name', 'state_name', 'population']]
                 .sort_values('population', ascending=False)
                 .head(10)
                 .sort_values('population', ascending=True)
                 .copy())
top_districts['district_label'] = (top_districts['district_name']
                                   + ' - ' + top_districts['state_name'])

plt.figure(figsize=(14, 8))
colors = sns.color_palette('mako', n_colors=len(top_districts))
ax = sns.barplot(
    data=top_districts,
    x='population',
    y='district_label',
    hue='district_label',
    palette=colors,
    legend=False,
)

for container in ax.containers:
    ax.bar_label(container, fmt='{:,.0f}', padding=6, fontsize=10, color='#263238')

ax.xaxis.set_major_formatter(FuncFormatter(lambda value, position: f'{value / 1_000_000:.1f}M'))
ax.set_title('Top 10 Districts by Population', fontsize=20, fontweight='bold', pad=18, color='#263238')
ax.set_xlabel('Population (millions)', fontsize=12, labelpad=10, color='#455A64')
ax.set_ylabel('District - State', fontsize=12, labelpad=10, color='#455A64')
ax.tick_params(axis='both', labelsize=10, colors='#455A64')
ax.grid(axis='x', linestyle='--', linewidth=0.8, alpha=0.35)
ax.grid(axis='y', visible=False)
ax.set_axisbelow(True)

sns.despine(left=True, bottom=True)
plt.tight_layout()
plt.show()
```

## **3. Create a doughnut chart showing the overall male and female population shares.**

```python
gender_population = pd.Series({
    'Male': df['male'].sum(),
    'Female': df['female'].sum(),
})

fig, ax = plt.subplots(figsize=(9, 7))
fig.patch.set_facecolor('#F7FAFC')
ax.set_facecolor('#F7FAFC')
colors = ['#2F6690', '#E76F51']

wedges, _, autotexts = ax.pie(
    gender_population,
    labels=None,
    colors=colors,
    startangle=90,
    counterclock=False,
    autopct='%1.1f%%',
    pctdistance=0.78,
    wedgeprops={'width': 0.38, 'edgecolor': '#F7FAFC', 'linewidth': 3},
    textprops={'color': 'white', 'fontsize': 12, 'fontweight': 'bold'},
)

ax.text(0, 0.08, f"{gender_population.sum():,.0f}", ha='center', va='center',
        fontsize=18, fontweight='bold', color='#1D3557')
ax.text(0, -0.10, 'Total Population', ha='center', va='center',
        fontsize=10, color='#607D8B')
ax.set_title('Overall Male and Female Population Shares', fontsize=19,
             fontweight='bold', pad=22, color='#1D3557')
ax.legend(
    wedges,
    [f'{label}: {value:,.0f}' for label, value in gender_population.items()],
    title='Population by Gender',
    loc='lower center',
    bbox_to_anchor=(0.5, -0.08),
    ncol=2,
    frameon=False,
    fontsize=10,
    title_fontsize=11,
)
ax.axis('equal')
plt.tight_layout()
plt.show()
```

## **4. Compare the 10 states with the highest literacy percentages.**

```python
state_literacy = (df.groupby('state_name', as_index=False)[['literate', 'population']]
                   .sum())
state_literacy['literacy_percentage'] = (
    state_literacy['literate'] / state_literacy['population'] * 100
)
top_literacy_states = (state_literacy
                       .nlargest(10, 'literacy_percentage')
                       .sort_values('literacy_percentage', ascending=True))

fig, ax = plt.subplots(figsize=(14, 8))
fig.patch.set_facecolor('#F7FAFC')
ax.set_facecolor('#F7FAFC')
colors = sns.color_palette('crest', n_colors=len(top_literacy_states))
bars = ax.barh(
    top_literacy_states['state_name'],
    top_literacy_states['literacy_percentage'],
    color=colors,
    edgecolor='white',
    linewidth=1.2,
)

ax.bar_label(bars, fmt='%.2f%%', padding=8, fontsize=10,
             color='#263238', fontweight='bold')
ax.set_title('Top 10 States by Literacy Percentage', fontsize=21,
             fontweight='bold', pad=20, color='#1D3557')
ax.text(0, 1.01, 'Literacy percentage = Literate population / Total population',
        transform=ax.transAxes, fontsize=11, color='#607D8B', va='bottom')
ax.set_xlabel('Literacy Percentage', fontsize=12, labelpad=10, color='#455A64')
ax.set_ylabel('State', fontsize=12, labelpad=10, color='#455A64')
ax.set_xlim(0, top_literacy_states['literacy_percentage'].max() + 5)
ax.xaxis.set_major_formatter(FuncFormatter(lambda value, position: f'{value:.0f}%'))
ax.tick_params(axis='both', labelsize=10, colors='#455A64', length=0)
ax.grid(axis='x', linestyle='--', linewidth=0.8, alpha=0.30, color='#90A4AE')
ax.grid(axis='y', visible=False)
ax.set_axisbelow(True)
sns.despine(left=True, bottom=True)
plt.tight_layout()
plt.show()
```

## **5. Compare male and female literacy percentages in the top 10 most populous states.**

```python
state_population = (df.groupby('state_name', as_index=False)
                    [['population', 'male', 'female', 'male_literate', 'female_literate']]
                    .sum())
top_populous_states = state_population.nlargest(10, 'population').copy()
top_populous_states['male_literacy_percentage'] = (
    top_populous_states['male_literate'] / top_populous_states['male'] * 100
)
top_populous_states['female_literacy_percentage'] = (
    top_populous_states['female_literate'] / top_populous_states['female'] * 100
)
top_populous_states = top_populous_states.sort_values('population', ascending=True)

fig, ax = plt.subplots(figsize=(16, 9))
fig.patch.set_facecolor('#F7FAFC')
ax.set_facecolor('#F7FAFC')
states = top_populous_states['state_name']
positions = np.arange(len(states))
bar_width = 0.36
male_bars = ax.bar(positions - bar_width / 2,
                   top_populous_states['male_literacy_percentage'],
                   width=bar_width, label='Male Literacy', color='#277DA1',
                   edgecolor='white', linewidth=1.2)
female_bars = ax.bar(positions + bar_width / 2,
                     top_populous_states['female_literacy_percentage'],
                     width=bar_width, label='Female Literacy', color='#F9844A',
                     edgecolor='white', linewidth=1.2)

ax.bar_label(male_bars, fmt='%.1f%%', padding=4, fontsize=9,
             color='#1D3557', fontweight='bold')
ax.bar_label(female_bars, fmt='%.1f%%', padding=4, fontsize=9,
             color='#8C3D20', fontweight='bold')
ax.set_title('Male vs Female Literacy in the Top 10 Most Populous States',
             fontsize=21, fontweight='bold', pad=18, color='#1D3557')
ax.text(0, 1.005, 'States are ranked by total population | Percentages are calculated separately by gender', transform=ax.transAxes,
        fontsize=11, color='#607D8B', va='bottom')
ax.set_xlabel('State', fontsize=12, labelpad=14, color='#455A64')
ax.set_ylabel('Literacy Percentage', fontsize=12, labelpad=12, color='#455A64')
ax.set_xticks(positions, states, rotation=30, ha='right')
ax.set_ylim(0, 108)
ax.yaxis.set_major_formatter(FuncFormatter(lambda value, position: f'{value:.0f}%'))
ax.tick_params(axis='both', labelsize=10, colors='#455A64', length=0, pad=8)
ax.grid(axis='y', linestyle='--', linewidth=0.8, alpha=0.30, color='#90A4AE')
ax.grid(axis='x', visible=False)
ax.set_axisbelow(True)
ax.legend(frameon=False, ncol=2, loc='upper right', bbox_to_anchor=(0.99, 0.99),
          fontsize=10, title='Literacy Group', title_fontsize=11,
          handlelength=1.4, columnspacing=1.5)
sns.despine(left=True, bottom=True)
plt.tight_layout(rect=[0, 0.04, 1, 0.92])
plt.show()
```

## **10. Create a single Python dashboard-style figure with four KPI summaries and two supporting charts.**

```python
dashboard_population = df['population'].sum()
dashboard_households = df['households'].sum()
dashboard_literacy = df['literate'].sum() / dashboard_population * 100
dashboard_worker_participation = df['workers'].sum() / dashboard_population * 100
dashboard_states = (df.groupby('state_name', as_index=False)['population'].sum()
                     .nlargest(10, 'population')
                     .sort_values('population', ascending=True))
dashboard_gender = pd.Series({'Male': df['male'].sum(), 'Female': df['female'].sum()})

fig = plt.figure(figsize=(18, 11), facecolor='#F7FAFC')
grid = fig.add_gridspec(3, 4, left=0.10, right=0.90, height_ratios=[0.9, 1.45, 2.2], hspace=0.48, wspace=0.30)
fig.suptitle('India District Census 2011 | Analytics Dashboard', fontsize=24, fontweight='bold', color='#1D3557', y=0.98)

kpis = [
    ('Total Population', f'{dashboard_population / 1_000_000:.1f}M', '#277DA1'),
    ('Total Households', f'{dashboard_households / 1_000_000:.1f}M', '#43AA8B'),
    ('Literacy Percentage', f'{dashboard_literacy:.2f}%', '#F9C74F'),
    ('Worker Participation', f'{dashboard_worker_participation:.2f}%', '#F9844A'),
]
for index, (label, value, color) in enumerate(kpis):
    card = fig.add_subplot(grid[0, index])
    card.set_facecolor('white')
    for spine in card.spines.values():
        spine.set_edgecolor(color)
        spine.set_linewidth(2)
    card.set_xticks([])
    card.set_yticks([])
    card.text(0.5, 0.62, value, ha='center', va='center', fontsize=23, fontweight='bold', color=color)
    card.text(0.5, 0.25, label, ha='center', va='center', fontsize=10, color='#455A64')

population_ax = fig.add_subplot(grid[1:, :2])
population_ax.set_facecolor('white')
population_bars = population_ax.barh(dashboard_states['state_name'], dashboard_states['population'], color=sns.color_palette('mako', 10), edgecolor='white')
population_ax.bar_label(population_bars, fmt='%.1fM', labels=[f'{value / 1_000_000:.1f}M' for value in dashboard_states['population']], padding=5, fontsize=9)
population_ax.set_title('Top 10 States by Population', fontweight='bold', color='#1D3557', pad=12)
population_ax.xaxis.set_major_formatter(FuncFormatter(lambda value, position: f'{value / 1_000_000:.0f}M'))
population_ax.set_xlabel('Population', color='#455A64')
population_ax.tick_params(axis='both', length=0, labelsize=9, colors='#455A64')
population_ax.grid(axis='x', linestyle='--', alpha=0.25)
population_ax.grid(axis='y', visible=False)
sns.despine(ax=population_ax, left=True, bottom=True)

gender_ax = fig.add_subplot(grid[1:, 2:])
gender_ax.set_facecolor('white')
gender_wedges, _, _ = gender_ax.pie(dashboard_gender, colors=['#277DA1', '#F9844A'], startangle=90, counterclock=False, autopct='%1.1f%%', pctdistance=0.76, wedgeprops={'width': 0.38, 'edgecolor': '#F7FAFC', 'linewidth': 3}, textprops={'color': 'white', 'fontsize': 12, 'fontweight': 'bold'})
gender_ax.set_title('Overall Gender Population Share', fontweight='bold', color='#1D3557', pad=12)
gender_ax.legend(gender_wedges, [f'{label}: {value / 1_000_000:.1f}M' for label, value in dashboard_gender.items()], frameon=False, loc='lower center', bbox_to_anchor=(0.5, -0.08), ncol=2, fontsize=9)
gender_ax.axis('equal')
plt.show()
```
