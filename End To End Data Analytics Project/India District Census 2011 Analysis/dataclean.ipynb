# **Data Analysis**

## **1. Which 10 states have the highest total population? Aggregate district records by state.**

```python
df.groupby(['state_name']).population.sum().sort_values(ascending=False).head(10)
```

#### **Uttar Pradesh, Maharashtra, Bihar, Andhra Pradesh, etc are the top 10 states with highest population.**

## **2. Which 10 districts have the highest population? Include their state names.**

```python
df.groupby(['district_name','state_name'],as_index=False).population.sum().sort_values(by = 'population',ascending=False).head(10)
```

## **3. Which five states have the highest and lowest female-to-male ratios?**

```python
df.groupby(['state_name']).sex_ratio.mean().round(2).sort_values(ascending=False).head(10)
```

## **4.  What is the overall literacy percentage (Literate / Population × 100) for each state? Rank states by percentage**


```python
df.groupby('state_name').literacy_percentage.sum()
```

## **6.  For each state, what are the total Workers and Non_Workers, and what percentage of the population are Workers?**

```python
df.head(2)
```

```python
df['worker_percentage'] = ((df.workers/df.population) * 100).round(2)
```

```python
df.groupby('state_name')[['workers','non_workers','worker_percentage']].sum()
```

## **7.  Which 10 districts have the highest share of agricultural workers (Agricultural_Workers / Workers × 100)?**

```python
df.head(2)
```

```python
df['agriculture_workers_share'] = ((df.agricultural_workers/df.workers)/100).round(2)
```

```python
df.groupby('district_name')['agriculture_workers_share'].sum().sort_values(ascending=False).head(10)
```

## **8.  Which states have the highest number of Households_with_Internet, and what is the percentage relative to Households?**

```python
df.head(5)
```

```python
df.groupby('state_name').households_with_internet.sum().sort_values(ascending=False).head(5)
```

```python
df['households_percentage'] = ((df.households_with_internet / df.households) * 100).round(2)
```

```python
df.groupby('state_name')[['households_with_internet','households_percentage']].sum().sort_values( by = 'households_with_internet',ascending=False).head(5)
```
