```python
import pandas as pd
import numpy as np
from sqlalchemy import create_engine
```

```python
df1 = pd.read_excel('Pizza_ingredients.xlsx')
df2 = pd.read_excel('Pizza_Sale.xlsx')
```

## **DATA CLEAN - PIZZA INGREDIENTS FILE**

```python
df1
```

```python
df1.info()
```

```python
df1.duplicated().sum()
```

```python
df1 = df1.rename(columns={'Items_Qty_In_Grams' : 'items_qty_in_grams'})
```

```python
df1
```

```python
df1.items_qty_in_grams = df1.items_qty_in_grams.fillna(df1.items_qty_in_grams.median())
```

```python
df1.isnull().sum()
```

```python
df1
```

```python
engine = create_engine('mysql+pymysql://root:1234@localhost/pizza_sales_db')
```

```python
df1.to_sql(name='pizza_ingredients',con=engine,if_exists='replace',index=False)
```
