```python
import pandas as pd
from sqlalchemy import create_engine
```

```python
df = pd.read_csv('tourism_data.csv')
```

```python
df
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.columns = df.columns.str.lower()
```

```python
df.isnull().sum()
```

```python
df.drop(columns='countryjoin',inplace=True)
```

```python
df.isnull().sum()
```

```python
df.drop(columns='count',inplace=True)
```

```python
df.gdp = df.gdp.fillna(df.gdp.median())
df.population = df.population.fillna(df.population.median())
```

```python
df
```

```python
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
df
```

```python
df = df.rename(columns={'way in':'way_in'})
```

```python
df
```

```python
df.info()
```

```python
engine = create_engine('mysql+pymysql://root:1234@localhost/tourismdb')
```

```python
df.to_sql(name='tourism',con=engine,if_exists='replace',index=False)
```
