```python
import pandas as pd
from sqlalchemy import create_engine
```

```python
df = pd.read_csv('BigBasket Products.csv')
```

```python
df
```

```python
df.info()
```

## *Dropping Index Column*

```python
df.drop(columns='index',inplace=True)
```

## *Checking Null Values*

```python
df.isnull().sum()
```

## *Handling Null Values*

```python
df.rating = df.rating.fillna(df.rating.median())
```

```python
df.rating.unique()
```

```python
df.dropna(subset = 'description', inplace=True)
```

```python
df
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
df
```

```python
df['Discount_per'] = (((df.market_price - df.sale_price)/df.market_price) * 100).round(2)
```

```python
df
```

```python
engine = create_engine('mysql+pymysql://root:1234@localhost/bigbasketdb')
```

```python
df.to_sql(name='bb_products',con=engine,if_exists='replace',index=False)
```
